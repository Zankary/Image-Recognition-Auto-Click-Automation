```python
import pyautogui
import time
import random
import glob
import os

from pyautogui import ImageNotFoundException

def encontrar(nome_arquivo_imagem, confianca=0.90, region=None):
    try:
        # Retorna o retângulo (left, top, width, height) da imagem
        regiao = pyautogui.locateOnScreen(nome_arquivo_imagem, confidence=confianca, region=region)
        if regiao:
            return True
        else:
            print(f"Não foi possível encontrar a imagem '{nome_arquivo_imagem}' na tela.")
            return False
        
    except ImageNotFoundException:
        #print(f"A imagem '{nome_arquivo_imagem}' não foi encontrada (confiança < {confianca*100:.0f}%).")
        return False
    except Exception as e:
        print(f"Ocorreu um erro: {e}")
        return False




def encontrar_e_clicar(nome_arquivo_imagem, confianca=0.85, region=None):
    """
    Tenta encontrar uma imagem na tela e clicar em um ponto aleatório dentro dela.
    """

    #print(f"Buscando por '{nome_arquivo_imagem}' com confiança de {confianca*100:.0f}%...")

    try:
        # Retorna o retângulo (left, top, width, height) da imagem
        regiao = pyautogui.locateOnScreen(nome_arquivo_imagem, confidence=confianca, region=region)

        if regiao:
            x, y, largura, altura = regiao
            # Gera coordenadas aleatórias dentro da área da imagem
            #ponto_x = random.randint(x, x + largura - 1)
            #ponto_y = random.randint(y, y + altura - 1)
            
            #pyautogui.moveTo(x, y)
            pyautogui.moveTo(x, y)
            time.sleep(.1)
            pyautogui.click(x, y)
            #print(f"Imagem encontrada e clicada em ponto aleatório: ({ponto_x}, {ponto_y})")
            return True
        else:
            print(f"Não foi possível encontrar a imagem '{nome_arquivo_imagem}' na tela.")
            return False

    except ImageNotFoundException:
        #print(f"A imagem '{nome_arquivo_imagem}' não foi encontrada (confiança < {confianca*100:.0f}%).")
        return False
    except Exception as e:
        print(f"Ocorreu um erro: {e}")
        return False



# --- Exemplo de uso ---
#time.sleep(5)
#encontrar_e_clicar(
#    r'C:\Users\joaov\Desktop\Codigos\Autoultimatum\images\t03_stormcaller_runes.png',
#    confianca=0.95
#)
```

```python
import pyautogui, time

def definir_regiao():
    print("Posicione o mouse no canto superior esquerdo e aguarde...")
    time.sleep(3)
    x1, y1 = pyautogui.position()
    print(f"Primeiro ponto: {x1}, {y1}")

    print("Agora posicione no canto inferior direito e aguarde...")
    time.sleep(3)
    x2, y2 = pyautogui.position()
    print(f"Segundo ponto: {x2}, {y2}")

    largura = x2 - x1
    altura = y2 - y1
    print(f"Região definida: ({x1}, {y1}, {largura}, {altura})")
    return (x1, y1, largura, altura)
```
