```python
import pyautogui
import time
import glob
import os
from pyautogui import ImageNotFoundException

def encontrar_e_clicar(nome_arquivo_imagem, confianca=0.8):
    """
    Tenta encontrar uma imagem na tela com o nível de confiança especificado 
    e clica no centro dela se for encontrada.

    :param nome_arquivo_imagem: O caminho (string) para o arquivo de imagem (ex: 'botao.png').
    :param confianca: O limiar (threshold) de confiança (float entre 0.0 e 1.0).
    :return: True se a imagem foi encontrada e clicada, False caso contrário.
    """
    print(f"Buscando por '{nome_arquivo_imagem}' com confiança de {confianca*100:.0f}%...")

    try:
        # locateCenterOnScreen retorna as coordenadas (x, y) do centro da imagem
        # ou levanta ImageNotFoundException se não for encontrada.
        coordenadas_centro = pyautogui.locateCenterOnScreen(
            nome_arquivo_imagem, 
            confidence=confianca
        )

        if coordenadas_centro:
            # Se encontrou, move o mouse e clica
            pyautogui.click(coordenadas_centro)
            print(f"Imagem encontrada e clicada em: {coordenadas_centro}")
            return True
        else:
            # Esta parte só seria alcançada em versões antigas do pyautogui que 
            # retornavam None em vez de levantar uma exceção. 
            # Mas é mantida como um fallback seguro.
            print(f"Não foi possível encontrar a imagem '{nome_arquivo_imagem}' na tela.")
            return False

    except ImageNotFoundException:
        # Exceção levantada pelo locateCenterOnScreen quando a imagem não é encontrada
        print(f"A imagem '{nome_arquivo_imagem}' não foi encontrada na tela (confiança < {confianca*100:.0f}%).")
        return False
    except Exception as e:
        print(f"Ocorreu um erro: {e}")
        return False

# --- Exemplo de Uso ---
# Certifique-se de que o arquivo 'seu_botao.png' existe no mesmo diretório 
# ou forneça o caminho completo.
# A confiança de 70% é 0.7.

time.sleep(5)
encontrar_e_clicar(r'C:\Users\joaov\Desktop\Codigos\Autoultimatum\images\t03_stormcaller_runes.png', confianca=0.95)
```

```python
DIRETORIO_IMAGENS = 'images'
caminhos_das_imagens = glob.glob(os.path.join(DIRETORIO_IMAGENS, '*.png'))
```

```python
time.sleep(3)

for i in caminhos_das_imagens:
    print('---')
    if not encontrar_e_clicar(i):
        print(f"Não achei {i}")

    else:
        print(f"Encontrei {i}")

    time.sleep(.5)
```
